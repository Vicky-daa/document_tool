```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
```

```python
os.getcwd()
```

```python
file_path = os.path.join(os.getcwd(),"data","sample.pdf")
```

```python
loader = PyPDFLoader(file_path=file_path)
```

```python
loader 
```

```python
document = loader.load()
```

```python
document
```

```python
len(document)
```

```python
document[0].metadata
```

```python
text_splitter = RecursiveCharacterTextSplitter( chunk_size = 500,
                                               chunk_overlap = 150,
                                               length_function = len)
```

```python
chunks = text_splitter.split_documents(document)
```

```python
len(chunks)
```

```python
chunks[0].metadata
```

```python

from langchain_google_genai import GoogleGenerativeAIEmbeddings

embedding_model=GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-2")

doc_vector=embedding_model.embed_query("What is the capital of France?")
```

```python
doc_vector
```

```python
from langchain_community.vectorstores import FAISS
```

```python
doc_vector=embedding_model.embed_documents(chunks[0].page_content)
```

```python
doc_vector
```

```python
vectorstore=FAISS.from_documents(chunks, embedding_model)
```

```python
relevant_doc=vectorstore.similarity_search("llama2 finetuning benchmark experiments.")
```

```python
relevant_doc
```

```python
relevant_doc=vectorstore.similarity_search("llama2 finetuning benchmark experiments.",k=10)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 10})
```

```python
retriever.invoke("llama2 finetuning benchmark experiments.")
```

```python

retriever
```

```python
from langchain_classic.prompts import PromptTemplate
```

```python
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context", "question"]
)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)
```

```python
result = rag_chain.invoke("tell  me about the llama2 finetuning benchmark experiments?")
```
